# Importing important libraries , modules and functions


```python
import pandas as pd                 
import numpy as np                   
import matplotlib.pyplot as plt      
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)
```

# Reading the Dataset

```python
#https://www.kaggle.com/datasets/ayeshaseherr/diamonds
df = pd.read_csv('../Data/diamonds.csv')
```

# Analysing the Data

```python
print("Dataset dimensions:", df.shape)
print("\nColumn names:")
print(df.columns.tolist())
print("\nData types and null values info:")
df.info()
print("\nSummary of numeric columns:")
df.describe()
```

# Selecting Numeric Columns

```python
numeric_columns = df.select_dtypes(include=np.number).columns.tolist()
print("Numeric columns:", numeric_columns)
```

# Visual Analysis of the Data

```python
#Bar graph
Plot_column=numeric_columns[4]
plt.hist(df[Plot_column].dropna(), bins=20, edgecolor="Black")
plt.title(f"Bar graph of {Plot_column}")
plt.xlabel(Plot_column)
plt.ylabel("Number of diamonds")
plt.grid(alpha=0.29)
plt.show()
```
